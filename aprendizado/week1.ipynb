## Funções & Loops

Exemplo resolvido

```python
def calcular_velocidade_orbital(altitude_km, g=6.674e-11, massa=5.972e24):
    raio = (6371 + altitude_km) * 1000
    return (g * massa / raio) ** 0.5 / 1000

for alt in [200, 408, 800]:
    v = calcular_velocidade_orbital(alt)
    print(f"{alt} km → {v:.2f} km/s")
```

1) Ordene satelites = [("Amazonia-1", 752), ("CBERS-4", 645), ("SGDC-1", 35786)] pela altitude usando lambda + sorted().

```python
satelites = [("Amazonia-1", 752), ("CBERS-4", 645), ("SGDC-1", 35786)]
ordenado = sorted(satelites, key=lambda s: s[1])
print("Satélites brasileiros ordenados por altitude:", ordenado)
```

2) Escreva uma função contar_passagens(dias, intervalo_horas) usando while que conta quantas passagens de satélite ocorrem em N dias, se ele passa a cada intervalo_horas.

```python
def contar_passagens(dias, intervalo_horas): # Define a função com dias e intervalo

    horas_totais = dias * 24 # Converte os dias em horas totais

    passagens, horas_passadas = 0, 0 # Inicializa os contadores em zero
    
    while horas_passadas + intervalo_horas <= horas_totais: # Loop enquanto couber o intervalo
        horas_passadas += intervalo_horas # Adiciona o intervalo ao tempo passado
        passagens += 1 # Conta mais uma passagem
    return passagens # Retorna o total de passagens calculadas

dias = 7

print(f"\nPassagens em {dias} dias: {contar_passagens(dias, 6)}")
```
