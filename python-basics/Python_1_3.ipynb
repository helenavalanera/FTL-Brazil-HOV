### Percorrendo Strings

* **Usando uma instrução While**


**Exemplo: Contando Vogais**

````python
def count_vowels_while(input_string: str) -> int:
    vowels = "aeiou"
    count = 0
    i = 0

    # Convert the string to lowercase to make the comparison case insensitive
    input_string = input_string.lower()

    while i < len(input_string):
        if input_string[i] in vowels:
            count += 1
        i += 1
    return count


# Test the functions
word = "HELLO"

print(count_vowels_while(word))
````

* **Usando uma instrução For**

````python
def count_vowels_for(input_string: str) -> int:
    vowels = "aeiou"
    count = 0

    # Convert the string to lowercase to make the comparison case insensitive
    input_string = input_string.lower()

    for char in input_string:
        if char in vowels:
            count += 1
    return count

word = "Abracadabra"
print(f"Number of vowels in '{word}' using for loop: {count_vowels_for(word)}")
````


Existem vários modos nos quais um arquivo pode ser aberto:

* "r" : Modo leitura (padrão). Abre o arquivo para leitura.
* "w" : Modo escrita. Abre o arquivo para escrita (e o cria se não existir). Se o arquivo já existir, ele trunca o arquivo.
* "a" : Modo anexar. Abre o arquivo para escrita, mas anexará ao final do arquivo se ele existir.
* "b" : Modo binário. Lê/Escreve arquivos em formato binário, usado especialmente para arquivos não textuais como imagens, executáveis, etc.

## **Escrevendo em Arquivos**

Para escrever em um arquivo, você normalmente o abrirá no modo **"write"** ou **"append"**.




**Modo Escrita:**

* Se o arquivo já existir, apaga seu conteúdo ou substitui o conteúdo antigo pelo novo conteúdo.

* Se o arquivo não existir, abri-lo no modo "write" criará um novo arquivo vazio.


````python
fname = basepath + 'ex.txt'

# Open the file with write mode. This will create a new file or overwrite an existing one.
file = open(fname, "w")

file.write("We love programming\n")
file.write("We are Frontier Tech Leaders!\n")

file.close()
````

````python
fh = open(fname , 'r')

fh.read()
````

````python
with open(fname , 'r') as fh:
  print(fh.read())
````

**Escrever em Arquivo Existente (Modo Append)**
* Se o arquivo já existir, abri-lo no modo "append" não alterará seu conteúdo. Anexa novo conteúdo ao final do arquivo, preservando os dados existentes.

* Se o arquivo não existir, abri-lo no modo "append" criará um novo arquivo vazio.


````python
# Define the file path
file_path = basepath + "zen.txt"

# Use the 'with' statement to ensure the file gets closed properly
with open(file_path, 'a') as file_handle:
    file_handle.write("\nThis is all true!")  # Append a new line and some text
````

````python
fh = open(file_path , 'r')
fh.read()
````

````python
# Ensure the file exists and has content
with open(file_path, 'a+') as file:
    # Move the cursor to the beginning of the file and print its content
    file.seek(0)
    print("Before appending:")
    print(file.read())

# Append text to the file
with open(file_path, 'a') as file:
    file.write("\nThis is the appended text.")

# Check the content of the file after appending
with open(file_path, 'r') as file:
    print("\nAfter appending:")
    print(file.read())
````

**Fechando um Arquivo**
Após realizar operações em um arquivo, é uma boa prática fechá-lo usando o método close():


**Usando a Instrução 'with'**
Uma forma mais pythônica de trabalhar com arquivos é usar a *instrução with*. Ela garante que o arquivo seja fechado adequadamente quando não for mais necessário:


````python
with open(file_path, "r") as file:
    content = file.read()
    print(content)
````

**Observação:** Não é necessário chamar explicitamente close() ao usar with.
