> Python é uma linguagem de programação orientada a objetos

**O que isso significa?**

Programação Orientada a Objetos (POO) é um paradigma de programação que se concentra no uso de objetos e classes para criar programas.

> Um objeto é um conjunto de variáveis e funções inter-relacionadas.
> As variáveis são frequentemente chamadas de ***propriedades (atributos)*** do objeto.
> As funções (métodos) são chamadas de comportamentos dos objetos.


**Por que usar Programação Orientada a Objetos?**

* A POO torna o código mais reutilizável e mais fácil de trabalhar em programas maiores.
* Os programas em POO evitam a repetição de código, pois uma classe pode ser definida uma vez e reutilizada muitas vezes.

 * Um programa é composto por um ou mais objetos trabalhando em conjunto — os objetos utilizam as capacidades uns dos outros.

## **1. Classes**

As classes definem as propriedades (atributos) e os comportamentos (métodos) do objeto.

> Pense em uma classe como um molde (blueprint) para criar objetos.



**Por exemplo:** Uma classe `Cachorro` seria composta por características compartilhadas por todos os cachorros, como raça e cor do pelo (características), e a capacidade de latir e sentar (comportamentos).



## **2. Instância**

Uma instância de uma classe é um objeto. Também é conhecida como objeto de classe ou instância de classe.

É o objeto real criado em tempo de execução.

Um objeto pode conter dados e código para manipular esses dados.

Por exemplo: o objeto Golden Retriever é uma instância da classe Cachorro.

## 3. Método

Métodos são as funções que usamos para descrever o comportamento dos objetos.

Eles são definidos dentro de uma classe.

>Um método é um procedimento associado a um objeto e, geralmente, também a uma mensagem. Um objeto consiste em dados de estado e comportamento; estes compõem uma interface, que especifica como o objeto pode ser usado. Um método é um comportamento de um objeto parametrizado por um usuário [fonte](https://en.wikipedia.org/wiki/Method_(computer_programming)).

Em linguagem natural, os métodos são verbos.


O Golden Retriever, como um Cachorro, tem a capacidade de latir. Ou seja, `latir()` é um dos métodos do Golden Retriever. Ele também poderia `sentar()`, `correr()`, `comer()`, etc.

Observação: Usar um método no programa afeta apenas um objeto específico. Por exemplo: todos os cachorros podem latir, mas você precisa de apenas um cachorro para fazer o latido.

**Comentário sobre a Saída**

Uma classe fornece três tipos de métodos:

* **Método inicializador (Construtor):** método `__init__` que inicializa os **atributos** do objeto.
 * É chamado automaticamente quando o objeto é criado.
 * Conhecido como construtor da classe.

* **Métodos mágicos**: Métodos especiais com dois sublinhados em ambos os lados.
 * Por exemplo, `__add__` e `__mul__` são usados para, respectivamente, somar e multiplicar objetos da mesma classe.

* **Métodos de instância**: Métodos que pertencem ao objeto criado.
  * Por exemplo, `l.append(4)` adiciona um elemento ao final da lista.

## Herança

Herança é um conceito fundamental na programação orientada a objetos em que uma nova classe herda atributos e comportamentos (métodos) de uma classe existente.

* Isso permite a reutilização de código e o estabelecimento de relações entre a classe pai (base) e a classe filha (derivada).

* A nova classe é frequentemente chamada de **"filha"** ou **"subclasse"**, enquanto a classe da qual ela herda é chamada de "pai" ou "classe base".

**Principais Benefícios:**

* **Reutilização:** Você não precisa escrever o mesmo código novamente; a classe filha herda todos os membros da classe pai.
* **Extensibilidade:** Você pode adicionar mais funcionalidades a uma classe filha além do que ela herda.
* **Hierarquia:** Estabelece uma forma de hierarquia em que todas as classes relacionadas compartilham uma base comum.
