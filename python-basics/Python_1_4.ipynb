#1.  **LISTAS EM PYTHON**

Listas em Python são coleções ordenadas que podem conter tipos de dados mistos. Elas são semelhantes a arrays em outras linguagens de programação, mas com mais capacidades.
* Listas são mutáveis, o que significa que seu conteúdo pode ser alterado após a criação.




````python
lists = [4 , 'names', ['king', 87]]
````

````python
print(lists)
````

## **Propriedades das Listas**

* Ordenadas: Mantêm a ordem de inserção dos dados

````python
# 1. Ordered: Lists maintain the order of data insertion
ordered_list = ['apple', 'banana', 'cherry']
print("Ordered List:", ordered_list[0])

````


* Mutável: A lista é mutável e podemos modificar itens


````python

# 2. Changeable: Lists are mutable
ordered_list[0] = 'mango'
print(ordered_list)
````

* Heterogênea: A lista pode conter dados de diferentes tipos


````python

# 3. Heterogeneous: Lists can contain data of different types
heterogeneous_list = ['apple', 100, 3.14, True, ["sub-list", "elements"]]
print("Heterogeneous List:", heterogeneous_list)
````

* Permite duplicatas: Permite dados duplicados

````python

# 4. Contains duplicates: Lists allow duplicate data
duplicate_list = ['apple', 'apple', 'banana', 'cherry', 'banana']
print("List with Duplicates:", duplicate_list)
````

## **Métodos de Listas**
Existem vários métodos que podem ser usados com listas para manipulá-las e inspecioná-las.
Alguns métodos comuns incluem
* append(),

````python
# Using the append method to add an item to the end of the list
duplicate_list = ['apple', 'apple', 'banana', 'cherry', 'banana']

duplicate_list.append('mango')
print(duplicate_list)
````

````python

# Using the extend method to add multiple items
duplicate_list.extend(['oranges', 'tomatoes'])

print(duplicate_list)
````

## Fatiamento de Listas

 * O formato básico é ```lista[início:fim]```, onde a fatia começará no índice início e irá até, mas não incluirá, o índice fim.

````python
# Demonstrating list slicing
numbers = [0, 1, 2, 3, 4, 5]
print(numbers[2:4])
````

## Listas e Laços

* Usamos um laço para acessar todos os elementos em uma lista.

* O laço for é especialmente útil com listas, permitindo percorrer cada item da lista.

````python
dishes = ['kebab', 'baklava', 'simit', 'kumpir', 'manti', 'lahmacun', 'kofte', 'pide', 'ayran', 'mezze']

for dish in dishes:
    print('I love ' + dish + ' very much')

````

**Enumerando uma lista**
* A função enumerate() rastreia o índice de cada item conforme percorre a lista

````python
dishes = ['kebab', 'baklava', 'simit', 'kumpir', 'manti', 'lahmacun', 'kofte', 'pide', 'ayran', 'mezze']

for index, dish in enumerate(dishes):

  print( 'Dish Number: ' + str(index) + ' ' + dish)
````

# **Dicionários**

Dicionários são uma forma de armazenar informações que estão conectadas de alguma maneira.

Dicionários são usados para armazenar valores de dados em pares chave:valor.


image.png

[Fonte da imagem](https://www.analyticsvidhya.com/blog/2021/06/working-with-lists-dictionaries-in-python/)


### Sintaxe Geral

```
dictionary_name = {key_1: value_1, key_2: value_2, key_3: value_3}
```



````python
# Creating a dictionary
student = {
    "name": "John",
    "age": 20,
    "course": "Computer Science"
}
print(student)
````

### **Características dos Dicionários**

* As chaves são sempre únicas, e não pode haver duplicatas.
* Não há índice no dicionário, ou seja, eles não são ordenados. A chave é o iterador padrão e é usada para recuperar o valor.
* Dicionários são mutáveis (modificáveis)

````python
#examples to demonstratte these properties

fruits_dict = {"apple": 1 , "oranges": 2, "mangos":3}

fruits_dict['apple'] = 4

print(fruits_dict)
````

````python
fruits_dict['mangos'] = 10
````

````python
print(fruits_dict)
````

## Métodos de Dicionários
Dicionários oferecem vários métodos que ajudam a acessar, modificar e gerenciar os pares chave:valor.
* Alguns métodos comumente usados incluem keys(), values() e items().

````python
# Using dictionary methods
print(fruits_dict.keys())
print(fruits_dict.values())
````

````python
print(fruits_dict.items())
````

### **Operações comuns com dicionários**
Existem algumas operações comuns que você vai querer fazer com dicionários.
* Adicionar novos pares chave-valor,
* Modificar informações no dicionário,
* Remover itens dos dicionários.

**Adicionando novos pares chave-valor**

````python
# Create an empty dictionary.
african_countries = {}

print(african_countries)


# Fill the dictionary with country-capital pairs.
african_countries['Nigeria'] = 'Abuja'


print(african_countries)
````

````python
african_countries['Kenya'] = 'Nairobi'
african_countries['Egypt'] = 'Cairo'
african_countries['South Africa'] = 'Pretoria'
african_countries['Ghana'] = 'Accra'

# Print out the countries and their capitals.
for country, capital in african_countries.items():
    print("\nCountry: %s" % country)
    print("Capital: %s" % capital)
````

````python
african_countries.items()
````

**Modificando valores em um dicionário**


````python
#modify one of the values in dictionary
african_countries['Cameroon'] = 'Yaunde'

print(african_countries)
````

````python
african_countries['Cameroon'] = 'Yaounde'
print(african_countries)
````

**Removendo pares chave-valor**

* Use a instrução `del` seguida pelo dicionário e sua chave.

````python
# Remove any country and its capital from the dictionary

del african_countries['Cameroon']

print(african_countries)

````

**Modificando chaves em um dicionário**
 Podemos alterar uma chave de dicionário em duas etapas:

* Crie uma nova chave e copie o valor para a nova chave.
* Exclua a chave antiga, o que também exclui o valor antigo.

````python
#examples
african_countries['Niger'] = african_countries['Nigeria'] # Abuja

del african_countries['Nigeria']


````

````python
print(african_countries)
````

### Dicionários e Laços

Dicionários podem ser percorridos usando laços, o que permite iterar sobre cada par chave:valor. Podemos fazer isso de 3 maneiras:
* Você pode percorrer todos os pares chave-valor;
* Você pode percorrer as chaves e extrair os valores das chaves que interessam;
* Você pode percorrer os valores.


**Percorrendo todos os pares chave-valor**

````python
# Iterating through a dictionary using a for loop

for country, capital in african_countries.items():
  print(country)
  print(capital)
````

**Percorrendo todas as chaves em um dicionário**

````python
my_dict = {'key_1': 'value_1',
    'key_2': 'value_2',
    'key_3': 'value_3',
    }

for key in my_dict.keys():
    print('Key: %s' % key)

#for key in my_dict:
#    print('Key: %s' % key)
````

**Percorrendo todos os valores em um dicionário**

````python
my_dict = {'key_1': 'value_1',
    'key_2': 'value_2',
    'key_3': 'value_3',
    }

for value in my_dict.values():
    print('Value: %s' % value)
````

# Tuplas

Tuplas são basicamente listas que nunca podem ser alteradas.

Tuplas em Python são coleções ordenadas e imutáveis. Isso significa que, uma vez que uma tupla é criada, seu conteúdo não pode ser alterado. Tuplas podem conter tipos de dados mistos e são semelhantes a listas, mas têm certas vantagens em contextos específicos devido à sua imutabilidade.


````python
# Creating a tuple
sample_tuple = (1, 2, "Python", 3.5)
print(sample_tuple)
````

````python

# convert from a list
tuple(['Veery', 'Eastern Meadowlark', 10, 25, True])
````

````python

# blank tuple
tuple()
````

**Acessando elementos**

````python
countries = list(african_countries.keys())

print(countries)
````

````python
c_tup = tuple(countries)
print(c_tup)
````

````python
c_tup[2]
````

**Fatiamento**

````python
c_tup[1:4]
````

### **Tuplas e Listas**

````python
# Converting between lists and tuples
sample_list = [1, 2, 3, 4]
converted_tuple = tuple(sample_list)
print(converted_tuple)

converted_list = list(converted_tuple)
print(converted_list)
````

### **Tuplas como Valores de Retorno**
* Tuplas são frequentemente usadas em funções para retornar múltiplos valores simultaneamente

````python
# Function that returns a tuple
def min_max(numbers):
    return (min(numbers), max(numbers))

result = min_max([3, 1, 4, 2, 5])
print(result)  # Outputs: (1, 5)
````

### **(Opcional) Dicionários e Tuplas**

````python
# Extracting keys and values as tuples and sorting
word_freq = {"Python": 3, "is": 3, "great": 1}
sorted_by_key = sorted(word_freq.items())
print(sorted_by_key)

# Sorting by values
sorted_by_value = sorted(word_freq.items(), key=lambda item: item[1], reverse=True)
print(sorted_by_value)
````
