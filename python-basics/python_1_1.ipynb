**Operadores de Comparação (retornam True ou False):**

* Igual a :  ==
*  Diferente de: !=
* Menor que :  <  
* Maior que:  >
* Menor ou igual a:  <=
*  Maior ou igual a:  >=

````python
x = 5
y = 10
print(x == y)  # False
print(x != y)  # True
print(x < y)   # True
print(x > y)   # False
print(x <= y)  # True
print(x >= y)  # False
````

**Operadores de Atribuição:**

Atribuir valor :

**Operadores Bit a Bit:**

* E bit a bit : &
* OU bit a bit: |
*  XOR bit a bit: ^
*  NÃO bit a bit: ~
*  Deslocamento à esquerda: <<
*  Deslocamento à direita: >>

````python
a = 60  # 60 = 0011 1100 in binary
b = 13  # 13 = 0000 1101 in binary
print(a & b)   # 12 = 0000 1100
print(a | b)   # 61 = 0011 1101
print(a ^ b)   # 49 = 0011 0001
print(~a)      # -61 (in two's complement form)
print(a << 2)  # 240 = 1111 0000
print(a >> 2)  # 15 = 0000 1111
````

# Nosso Primeiro Programa Real (Planejando uma Viagem de Carro)

**Nota:** Os computadores basicamente fazem três coisas: entrada, processamento e saída

**Problema:** Calcular o Custo Total de uma Viagem de Carro

**Cenário:** Você está planejando uma viagem de carro. Para fazer o orçamento da viagem, você deseja calcular o custo total de combustível para todo o percurso. Você tem as seguintes informações:

A distância total da viagem em quilômetros.
A eficiência de combustível do seu carro em quilômetros por litro.
O custo do combustível por litro.

**Objetivo:**
 Calcular o custo total de combustível para a viagem.*

````python
# Variables provided for the scenario

# Total distance of the trip in miles
distance = 500  # e.g., 500 km

# Fuel efficiency of the car (miles per gallon)
fuel_efficiency = 25  # e.g., 25 mpg. This means the car can travel 25 miles with 1 gallon of fuel.

# Cost of fuel per gallon
cost_per_gallon = 3.50  # e.g., $3.50 per gallon

# Using expressions to calculate the total fuel required and the total cost

# Total fuel required for the trip (distance / fuel efficiency)
total_fuel = distance / fuel_efficiency  # Using the division operator to determine gallons of fuel required

# Total cost of fuel for the trip (total fuel * cost per gallon)
total_cost = total_fuel * cost_per_gallon  # Using the multiplication operator to calculate the cost

# Displaying the result
print(f"The total cost of fuel for the trip is ${total_cost:.2f}")  # Using an f-string to format the result

# Output: The total cost of fuel for the trip is $70.00
````

## Instruções Condicionais

As instruções condicionais na programação são cruciais porque permitem que um programa tome decisões e execute diferentes caminhos de código com base em certas condições.

Assim como os humanos tomam decisões todos os dias sobre que roupa vestir dependendo das condições climáticas, quando atravessar a rua com base na cor do semáforo, participar deste bootcamp de Python dependendo se você tomou café da manhã ou não, etc.


Em Python, as principais instruções condicionais são introduzidas usando if, elif e else

 Basicamente, as instruções condicionais testam se uma condição é True ou False.

**Por que Instruções Condicionais?**

**Tomada de Decisão:** Assim como na vida real, onde tomamos decisões com base em certas condições (por exemplo, se estiver chovendo, levamos um guarda-chuva), os programas muitas vezes precisam decidir qual ação tomar com base em circunstâncias específicas. As instruções condicionais permitem esse processo de tomada de decisão.

**Comportamento Dinâmico:** Elas permitem que o programa se adapte e produza resultados diferentes dependendo da entrada ou de outros fatores. Sem condicionais, os programas sempre seguiriam o mesmo caminho e forneceriam a mesma saída.

**Tratamento de Erros:** As verificações condicionais podem ajudar um programa a identificar situações inesperadas ou erros e tratá-los adequadamente, prevenindo falhas ou resultados incorretos.

**Otimização:** Usando instruções condicionais, um programa pode escolher a maneira mais eficiente de realizar uma tarefa com base nas condições atuais.

**Operadores de Comparação**
Em Python, os operadores de comparação são usados para comparar valores e retornam True ou False. Esses operadores desempenham um papel central nas instruções condicionais, especificamente na instrução if, pois determinam se um bloco de código específico deve ser executado.

As instruções if, elif e else em Python são fundamentais para os processos de tomada de decisão. Elas permitem verificações sequenciais e executam blocos de código específicos com base nas condições.


* **instrução if:** Começa com a palavra-chave if, seguida por uma condição e dois pontos (:). O bloco de código a ser executado se a condição for True é indentado abaixo do if.

* **instrução elif:** Abreviação de "else if". Permite condições adicionais após o if inicial. Vem depois de um if ou outro elif, seguido por uma condição e dois pontos (:).

* **instrução else:** Esta não tem condição. Ela captura qualquer coisa que não foi capturada pelas instruções if ou elif anteriores.

**Estrutura Básica**

````python
if condition1:
    # code block for condition1
elif condition2:
    # code block for condition2
else:
    # code block if none of the conditions are met
````

**Exemplo 1: Uma instrução if-else simples.**

````python
age = 17
if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
````

**Cenário:** Um cinema tem a seguinte política de preços de ingressos:

* Crianças com menos de 10 anos pagam R\$ 5.
* Adolescentes entre 10 e 17 anos pagam R\$ 7.
* Adultos entre 18 e 59 anos pagam R\$ 10.
* Idosos, com 60 anos ou mais, pagam R\$ 8.

Nas quartas-feiras, todos recebem um desconto de R$ 2, independentemente da faixa etária.
Dada a idade da pessoa e o dia da semana, determine o preço do ingresso.

````python
# Input age and day of the week
age = int(input("Enter your age: "))
day = input("Enter the day of the week: ").lower()

# Determine the base ticket price
if age < 10:
    price = 5
elif age <= 17:
    price = 7
elif age <= 59:
    price = 10
else:
    price = 8

# Apply discount if it's Wednesday
if day == "wednesday":
    price -= 2

# Display the final ticket price
print(f"The ticket price is ${price}.")
````
