### **Passo 1: Criar uma tabela**

Primeiro, configuraremos uma tabela chamada 'books' (livros) que armazenará o título, autor e ano de publicação dos livros.

```python
import sqlite3

# Connect to a database (or create one if it doesn't exist)
conn = sqlite3.connect('library.db')
c = conn.cursor()

# Create a table named 'books'
c.execute('''
CREATE TABLE IF NOT EXISTS books (
    id INTEGER PRIMARY KEY,
    title TEXT,
    author TEXT,
    year INTEGER
)
''')
#save all the changes made in database
conn.commit()
#close connection
conn.close()
```

# **Introdução aos Relacionamentos entre Tabelas**

* No mundo real, os dados frequentemente possuem relacionamentos.

 * Por exemplo, um estudante pode se matricular em vários cursos, e cada curso pode ter vários estudantes.

* No mundo dos bancos de dados digitais, relacionamentos são associações significativas entre tabelas que contêm informações relacionadas.

* Sem alguma conexão entre tabelas em um banco de dados, seria como trabalhar com arquivos de planilha separados em vez de um sistema de banco de dados.



**Relembrando:**
* Bancos de dados são coleções de tabelas, e essas tabelas possuem campos (colunas) e registros (linhas)

* Bancos de dados relacionais usam chaves para estabelecer relacionamentos (conexões) entre tabelas. Também para garantir a integridade dos dados e evitar redundância.

* Ao informar ao seu banco de dados que os valores de chave em uma tabela correspondem aos valores de chave em outra, você cria um relacionamento entre essas tabelas.

* Esses relacionamentos possibilitam a execução de consultas poderosas entre diferentes tabelas do seu banco de dados.

## **Compreendendo as Chaves em Tabelas de Bancos de Dados Relacionais**



[Image Source](https://www.google.com/url?sa=i&url=https%3A%2F%2Fmacxima.medium.com%2Fprimary-key-vs-foreign-key-in-sql-d89260b74e1c&psig=AOvVaw198H0vUZnEOCKXXJidVF13&ust=1693107827611000&source=images&cd=vfe&opi=89978449&ved=0CBIQjhxqFwoTCKjAq6i0-YADFQAAAAAdAAAAABAJ)

Três chaves fundamentais dos bancos de dados: chave primária, chave estrangeira e chave única.

### **1. Chave Primária (Primary Key)**

* Um identificador único para cada registro em uma tabela. Não são permitidas duplicatas ou valores nulos.
* Cada tabela contém uma chave primária (entidade), que identifica as linhas dentro dessa tabela.
* Por exemplo, cada estudante em uma escola pode ter um "ID do Estudante" específico que nenhum outro estudante possui. Este "ID do Estudante" pode servir como nossa chave primária.



### **2. Chave Estrangeira (Foreign Key)**

* Um campo em uma tabela que identifica exclusivamente um registro em outra tabela, estabelecendo assim um relacionamento entre elas.

* Quando a chave de entidade de uma tabela é vinculada a uma segunda tabela, ela é conhecida como chave estrangeira nessa segunda tabela.


>Quando uma chave primária aparece em outra tabela, esse campo é chamado de chave estrangeira nessa tabela.
>O relacionamento entre chaves primárias e chaves estrangeiras cria o relacionamento entre tabelas.

### **3. Chave Única (Unique Key)**

* Garante que o valor seja único em toda a tabela (onde foi definido).
* Diferentemente das chaves primárias, as chaves únicas podem ter valores nulos (ou vazios), e uma tabela pode ter mais de uma chave única.


**Observação:**

* Identificar as conexões que você precisará entre as tabelas faz parte do processo de modelagem de dados e design de esquema (ver adiante)

## **Relacionamentos em Tabelas**


* Para representar relacionamentos em bancos de dados, usamos os seguintes termos:


## **1. Relacionamento Um-para-Um**



**Exemplo do mundo real:**

>Todo ser humano tem um ID único dependendo do país. Nesta analogia, uma pessoa se relaciona a um e apenas um número de ID, e vice-versa.

**Contexto do banco de dados:**
>Este relacionamento implica que um registro em uma tabela (Tabela A) se relaciona a apenas um registro em outra tabela (Tabela B), e um registro na Tabela B se relaciona a apenas um registro na Tabela A.

**Exemplo:**
>Considere duas tabelas: Usuários e Licenças. Cada usuário tem apenas uma carteira de motorista, e cada carteira de motorista pertence a apenas um usuário. Aqui, um registro na tabela Usuários se relaciona a um único registro na tabela Licenças, estabelecendo um relacionamento um-para-um.

* Relacionamentos um-para-um também podem ser úteis para fins de segurança. Por exemplo: Você pode querer armazenar informações confidenciais de clientes em uma tabela separada, vinculada à sua tabela principal de Clientes por meio de uma chave estrangeira.



## **2. Relacionamento Um-para-Muitos:**

**Exemplo do mundo real:**
>Uma mãe pode ter vários filhos, mas cada filho tem uma única mãe biológica. Aqui, uma mãe se relaciona a muitos filhos, mas cada filho se relaciona a uma mãe.

**Contexto do banco de dados:**
>Um registro na Tabela A pode se relacionar a muitos registros na Tabela B, mas um registro na Tabela B se relaciona a apenas um registro na Tabela A.


**Exemplo:**
>Considere duas tabelas: Autores e Livros. Um autor pode escrever muitos livros, mas cada livro tem apenas um autor principal. Neste caso, um registro na tabela Autores pode se relacionar a vários registros na tabela Livros, estabelecendo um relacionamento um-para-muitos.


* Relacionamentos um-para-muitos são o tipo mais comum de relacionamentos entre tabelas em um banco de dados.



##3. **Relacionamento Muitos-para-Muitos:**

**Exemplo do mundo real:**
>Em uma escola, um estudante pode se matricular em vários cursos, e um curso pode ter vários estudantes. Um estudante se relaciona a muitos cursos, e um curso se relaciona a muitos estudantes.

**Contexto do banco de dados:**
>Registros na Tabela A podem se relacionar a vários registros na Tabela B, e vice-versa.

**Exemplo:**
>Considere duas tabelas: Estudantes e Cursos. Como um estudante pode se matricular em vários cursos e um curso pode ter vários estudantes, existe um relacionamento muitos-para-muitos entre eles. No entanto, bancos de dados relacionais não podem representar diretamente relacionamentos muitos-para-muitos. Assim, introduzimos uma terceira tabela, frequentemente chamada de tabela de junção ou ponte, para resolver isso.

Para o exemplo de estudantes e cursos, a terceira tabela poderia ser CursosEstudantes com dois campos principais: ID_Estudante e ID_Curso. Cada combinação de ID_Estudante e ID_Curso representa um estudante matriculado em um curso específico. Esta tabela ajuda a dividir o relacionamento muitos-para-muitos em dois relacionamentos um-para-muitos: um entre Estudantes e CursosEstudantes, e outro entre Cursos e CursosEstudantes.

# **Projetando um Banco de Dados Relacional [Estudo de Caso - Varejista]**

Percorreremos as etapas de design de um banco de dados para um varejista que deseja gerenciar seus produtos, clientes, pedidos e funcionários.

**Objetivo:**

> Projetar um banco de dados para um novo varejista na cidade.

**Cenário**
> Este varejista se preocupa muito com o relacionamento com os clientes e deseja recompensar os clientes que atingirem uma meta de gastos e presentear esses clientes principais no aniversário de 1 ano da primeira compra. Este varejista precisa de uma maneira de organizar os produtos por preço e categoria para fazer recomendações inteligentes aos clientes com base na idade. Este varejista também deseja acompanhar os funcionários de melhor desempenho para recompensar aqueles com as maiores vendas com um aumento no final do ano.

* Projetar um banco de dados relacional envolve uma abordagem estruturada para garantir que os dados estejam bem organizados, mantenham sua integridade e possam ser acessados de forma eficiente.

* O processo de design envolve modelagem de dados e design de esquema

* O modelo de dados é a representação inicial de alto nível dos requisitos e relacionamentos dos dados.
 * Ele fornece uma visão abstrata dos elementos de dados, seus relacionamentos e outras restrições sem aprofundar em detalhes específicos do sistema de banco de dados.

* O esquema é a estrutura que definimos para nossos dados. O esquema define as tabelas, relacionamentos entre tabelas, campos e índices.




* O processo de modelagem de dados e design de esquema permite que você descubra como seus dados se encaixam e como exatamente você deve configurar suas tabelas e seus campos.


* Uma ferramenta que nos ajudará a projetar nosso esquema é um DER (diagrama entidade-relacionamento).
 * Usaremos o [Lucidchart para construir nosso DER](https://www.lucidchart.com/pages/), e você pode se cadastrar gratuitamente.

* Este diagrama nos permitirá visualizar nossas entidades e seus relacionamentos.

Aqui estão as principais tarefas ao projetar nosso esquema:

* Compreender as necessidades do negócio
* Identificar entidades
* Identificar propriedades/campos nessas entidades
* Definir relacionamentos entre tabelas

### **Passo 1: Compreender as Necessidades do Negócio**

* Ao projetar um esquema de banco de dados relacional, o primeiro passo é compreender as necessidades do negócio.
* Isso nos ajuda a determinar que tipo de informação devemos armazenar.
  * Por exemplo, se estamos trabalhando com um varejista que deseja oferecer um presente de aniversário para clientes no primeiro aniversário de compra, precisaríamos armazenar a data em que o cliente se associou.

Um resumo dos requisitos para nossos clientes:

* Armazenar o total de gastos do cliente até a data atual
* Armazenar a data de aniversário da primeira compra do cliente
* Armazenar a idade do cliente
* Armazenar o total de vendas dos funcionários em valor monetário
* Armazenar produtos e incluir categoria e propriedade de preço

### **Passo 2: Definindo Entidades (Tabelas)**

* O próximo passo é definir as entidades sobre as quais queremos armazenar dados.

* No nosso caso do varejista, as entidades devem ser:

  * clientes (customers)
  * produtos (products)
  * pedidos (orders)
  * funcionários (employees)

* Usando a ferramenta DER, nomeamos cada entidade.

> Observação: Isso poderia ser expandido para adicionar mais entidades, como lojas se houver múltiplas localizações, fabricantes, etc., dependendo das necessidades do negócio.

image.png

### **Passo 3: Definindo Propriedades**

* Uma vez que identificamos as entidades, devemos definir quais campos queremos armazenar sobre essas entidades.
* Observação: cada tabela, ou entidade, deve ter uma propriedade única de identificação **(chave primária)**.
 * Por exemplo, podemos ter clientes com o mesmo nome ou mesma data de nascimento.

* Duas maneiras comuns de criar uma chave primária:
 * Gerar programaticamente um valor único
 * Atribuir um número inteiro que aumenta automaticamente a cada novo registro





image.png

* Tudo nas propriedades das tabelas vem diretamente das especificações que o negócio nos forneceu.
 * Por exemplo, o negócio quer saber qual cliente fez a compra, qual funcionário realizou a venda e quais produtos estavam no pedido.
 Observe que na tabela Pedidos (Orders), um customerID, employeeID e productID são referenciados para atender a essas necessidades.

### **Passo 4: Definir Relacionamentos**

* Após identificar entidades e suas propriedades, pense em como essas tabelas se relacionam entre si.
* Uma tabela pai terá uma coluna de chave primária única, e uma tabela filha terá sua própria chave primária e, em seguida, uma coluna parent_id que referencia a tabela pai.


image.png


>Por exemplo, a tabela de clientes (customers) tem um customerID, que é a chave primária. Na tabela Pedidos (Orders), definimos um orderID como chave primária e referenciamos o customerID para indicar qual cliente fez o pedido. Da mesma forma, também temos uma coluna referenciando a tabela Funcionários (Employees), employeeID, para indicar qual funcionário realizou a venda.
>Quando uma chave primária aparece em outra tabela, esse campo é chamado de chave estrangeira nessa tabela. O relacionamento entre chaves primárias e chaves estrangeiras cria o relacionamento entre tabelas.
