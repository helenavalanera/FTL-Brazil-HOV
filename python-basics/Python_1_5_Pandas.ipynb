-  Carregando Dados de Monitoramento Ambiental

Vamos criar um dataset simulado de estações de monitoramento de desmatamento na Amazônia. Este é o tipo de dado que você encontraria em projetos reais de sensoriamento remoto.

````python
np.random.seed(42)
n = 200

# Criando um DUMMY dataset de alertas de desmatamento
df = pd.DataFrame({
    "data": pd.date_range("2023-01-01", periods=n, freq="W"),
    "latitude": np.random.uniform(-12.0, -2.0, n),
    "longitude": np.random.uniform(-65.0, -50.0, n),
    "area_desmatada_ha": np.random.exponential(50, n),
    "tipo_alerta": np.random.choice(
        ["DETER", "PRODES", "GLAD"],
        n,
        p=[0.5, 0.3, 0.2]
    ),
    "confiabilidade": np.random.uniform(0.6, 1.0, n),
    "bioma": np.random.choice(
        ["Amazônia", "Cerrado", "Pantanal"],
        n,
        p=[0.7, 0.2, 0.1]
    )
})

print(f"Shape do dataset: {df.shape}")
df.head(10)
````

## 4. Filtragem e Seleção de Dados

Em análise espacial, frequentemente filtramos dados por região geográfica, período temporal ou limiares.

````python
# Filtrar apenas alertas da Amazônia
amazonia = df[df["bioma"] == "Amazônia"]
print(f"Alertas na Amazônia: {len(amazonia)}")

````

````python
# Filtrar por confiabilidade alta (> 0.9) E área grande (> 100 ha)
alerta_critico = df[(df["confiabilidade"] > 0.9) & (df["area_desmatada_ha"] > 100)]
print(f"Alertas críticos (alta confiança, >100ha): {len(alerta_critico)}")
print()
````

````python

# Filtrar por região geográfica (bounding box)
bbox = df[
    (df["latitude"] > -8.0) & (df["latitude"] < -4.0) &
    (df["longitude"] > -58.0) & (df["longitude"] < -54.0)
]
print(f"Alertas na região [-8, -4, -58, -54]: {len(bbox)}")
````

````python

# Filtro por data
junho_2023 = df[df["data"].dt.month == 6]
print(f"\nAlertas em junho de 2023: {len(junho_2023)}")
````

### Usando query() e loc/iloc

````python
# query() — sintaxe expressiva para filtrar
resultado = df.query("area_desmatada_ha > 200 and confiabilidade > 0.95")
print(f"Áreas >200ha com confiabilidade >0.95: {len(resultado)}")

# loc — acesso por label
print("\nPrimeiras 3 linhas, colunas específicas:")
print(df.loc[:2, ["data", "bioma", "area_desmatada_ha"]])

# iloc — acesso por posição
print("\nLinhas 10-12, colunas 0-3:")
print(df.iloc[10:13, 0:4])
````

loc vs iloc — seleção por nome (label) ou por posição

Uma das diferenças mais importantes do pandas é entender quando usar loc e quando usar iloc.

A ideia principal é:

- .loc : seleciona usando os labels (nomes/rótulos)

- .iloc : seleciona usando posições inteiras (0, 1, 2, ...)

Ambos permitem selecionar linhas e colunas ao mesmo tempo.

````python
df.loc[ df["bioma"] == "Amazônia" ]

df.loc[ df["bioma"] == "Amazônia", ["data", "bioma", "area_desmatada_ha"] ]
````

````python
df.loc[df['area_desmatada_ha']>150]
````

## 5. Agregação e GroupBy

Agrupar dados por categoria (bioma, tipo de alerta, período) é essencial para análise espacial.

````python
# Agrupar por bioma — estatísticas de desmatamento
print("=== Desmatamento por bioma ===")
grp_bioma = df.groupby("bioma").agg(
    total_alertas=("area_desmatada_ha", "count"),
    area_total=("area_desmatada_ha", "sum"),
    area_media=("area_desmatada_ha", "mean"),
    area_maxima=("area_desmatada_ha", "max"),
    confiabilidade_media=("confiabilidade", "mean")
).round(2)
print(grp_bioma.to_string())
print()

# Agrupar por tipo de alerta
print("=== Desmatamento por tipo de alerta ===")
grp_tipo = df.groupby("tipo_alerta").agg(
    total=("area_desmatada_ha", "count"),
    area_media=("area_desmatada_ha", "mean"),
    area_total=("area_desmatada_ha", "sum")
).round(2)
print(grp_tipo.to_string())
````

## 6. Séries Temporais

Dados geoespaciais frequentemente têm componente temporal. Pandas tem excelente suporte para trabalhar com datas.

````python
# Definindo a coluna de data como índice
df_time = df.set_index("data")

# Agregando por mês
mensal = df_time.resample("ME").agg(
    numero_alertas=("area_desmatada_ha", "count"),
    area_total=("area_desmatada_ha", "sum"),
    area_media=("area_desmatada_ha", "mean")
).round(2)

print("=== Desmatamento mensal ===")
print(mensal.head(6).to_string())

# Mês com maior área desmatada
mes_pico = mensal["area_total"].idxmax()
print(f"\nMês com maior desmatamento: {mes_pico.strftime('%B/%Y')}")
print(f"Área total: {mensal.loc[mes_pico, 'area_total']:.1f} ha")
````

````python
# Agregando por trimestre
trimestral = df_time.resample("QE").agg(
    alertas=("area_desmatada_ha", "count"),
    area_total=("area_desmatada_ha", "sum")
).round(2)

print("=== Desmatamento trimestral ===")
print(trimestral.to_string())

# Tendência — média móvel de 4 semanas
df_time["media_movel_4s"] = df_time["area_desmatada_ha"].rolling(4).mean()
print(f"\nMédia móvel (últimas 5 semanas):")
print(df_time[["area_desmatada_ha", "media_movel_4s"]].tail(5).round(2).to_string())
````

## 7. Combinando Datasets — Merge e Join

Em SIG, frequentemente juntamos dados de diferentes fontes: shapefiles com tabelas censitárias, coordenadas com dados climáticos.

````python
# Criando dataset complementar com dados climáticos
np.random.seed(99)
clima = pd.DataFrame({
    "data": pd.date_range("2023-01-01", periods=52, freq="W"),
    "precipitacao_mm": np.random.exponential(15, 52),
    "temperatura_c": np.random.normal(27, 3, 52)
})

# Merge — juntando dados de desmatamento com dados climáticos
# Agregando desmatamento por semana primeiro
desmat_semanal = df.groupby("data").agg(
    area_total=("area_desmatada_ha", "sum"),
    n_alertas=("area_desmatada_ha", "count")
).reset_index()

# Merge (inner join por data)
merged = pd.merge(desmat_semanal, clima, on="data", how="inner")
print(f"Dataset combinado: {merged.shape}")
print(merged.head(8).round(2).to_string())
````

````python
# Criando um dataset de referência de municípios
municipios = pd.DataFrame({
    "municipio_id": [1, 2, 3, 4, 5],
    "nome": ["Altamira", "São Félix do Xingu", "Porto Velho", "Lábrea", "Novo Progresso"],
    "estado": ["PA", "PA", "RO", "AM", "PA"],
    "area_km2": [159534, 84213, 34090, 68230, 38162]
})
print("Municípios de referência:")
print(municipios.to_string())

# Simulando dados de desmatamento agregados por município
alertas_municipio = pd.DataFrame({
    "municipio_id": [1, 1, 1, 2, 2, 3, 4, 5, 5, 5],
    "ano": [2022, 2023, 2024, 2023, 2024, 2024, 2024, 2022, 2023, 2024],
    "area_desmatada_ha": [1200, 1500, 980, 800, 2100, 3500, 2700, 600, 1100, 1800]
})

# Agregando e juntando com dados do município
resumo = alertas_municipio.groupby("municipio_id").agg(
    total_desmatado=("area_desmatada_ha", "sum"),
    media_anual=("area_desmatada_ha", "mean")
).round(1).reset_index()

final = pd.merge(municipios, resumo, on="municipio_id", how="left")
final["taxa_desmatamento"] = (final["total_desmatado"] / final["area_km2"]).round(2)

print("\n=== Resumo de desmatamento por município ===")
print(final.to_string())
````

## 8. Criando Novas Colunas e Transformações

Frequentemente precisamos derivar novas informações dos dados existentes.

````python
# Adicionando colunas calculadas
df["mes"] = df["data"].dt.month
df["ano"] = df["data"].dt.year
df["estacao"] = df["mes"].apply(
    lambda m: "Seca" if m in [6, 7, 8, 9, 10] else "Chuvosa"
)

# Classificando severidade do desmatamento
def classificar_severidade(area):
    if area < 30:
        return "Baixa"
    elif area < 100:
        return "Média"
    elif area < 200:
        return "Alta"
    else:
        return "Crítica"

df["severidade"] = df["area_desmatada_ha"].apply(classificar_severidade)

print("Dataset com novas colunas:")
print(df[["data", "area_desmatada_ha", "estacao", "severidade"]].head(10).to_string())
print()

# Comparando estações
print("=== Desmatamento por estação ===")
print(df.groupby("estacao").agg(
    total_alertas=("area_desmatada_ha", "count"),
    area_media=("area_desmatada_ha", "mean"),
    area_total=("area_desmatada_ha", "sum")
).round(2).to_string())
````
