## **O que são Funções?**

Em programação, uma função é um *bloco de código autocontido* (conjunto de ações ou instruções) que encapsula uma *tarefa específica* ou um *grupo de tarefas* relacionadas.

> Você pode pensar em uma função como um mini-programa dentro do seu programa.

## **Por que precisamos de funções?**


Funções oferecem vários benefícios, incluindo:

* **Modularidade:** Funções permitem dividir tarefas complexas em partes menores e mais gerenciáveis.
* **Reutilização:** Uma vez que você escreveu uma função para realizar uma tarefa específica, você pode chamá-la quantas vezes precisar sem reescrever o mesmo código.

* **Abstração:** Funções ocultam a complexidade das tarefas que realizam.

* **Manutenibilidade:** Alterações no código são mais fáceis quando a funcionalidade está encapsulada em funções.

* **Teste e Depuração:** Funções podem ser testadas individualmente, o que facilita identificar e corrigir problemas.

## Funções Lambda
* Uma função lambda é uma pequena função anônima.
* "Anônima" significa que ela não tem um nome como as funções regulares definidas com def.
* Funções lambda são normalmente usadas para operações simples que podem ser expressas em uma única linha.
* Uma função lambda pode receber qualquer número de argumentos, mas só pode ter uma expressão.

image.png

````python
#Multiplication of two arguments which are a and b, return the result:

x = lambda a, b : a * b


#call lambda
print(x(10, 5))
````

**Comentário**:

* `lambda:` Esta palavra-chave indica que estamos definindo uma função lambda.
* a, b: Estes são parâmetros da função lambda, exatamente como parâmetros de função. A função lambda espera duas entradas quando chamada.
*

##  **Loop While**


O loop while executa um bloco de código enquanto sua **condição** for **True** (verdadeira).
É essencial modificar a variável de controle do loop dentro do loop para garantir que a condição do loop eventualmente se torne **False** (falsa), caso contrário você terá um **loop infinito**.

* O loop while é como uma instrução if que se repete

* Loops têm variáveis de iteração que mudam a cada passagem pelo loop



**A Sintaxe Básica**
```
while condição:
    # código a executar enquanto a condição for True
```
* No loop while, a expressão/condição é verificada primeiro.
* O corpo do loop é executado apenas se a expressão/condição for avaliada como True.
* Após uma iteração, a expressão/condição é verificada novamente. Este processo continua até que a expressão de teste seja avaliada como False.

**Ponto-Chave**:
O loop while é usado quando:

* O número de iterações é determinado por uma condição em vez de uma contagem predeterminada.
* Você quer executar o loop indefinidamente até que uma determinada condição seja atendida.


**Exemplo 1**

````python
# Example 1: Print numbers less than 8
count = 1

while count < 8:

  print(count)

  count = count + 1

#code continues
print('The while is over')

````

**Exemplo 2:**
Aqui está um exemplo simples, mostrando como um jogo permanecerá ativo enquanto o jogador tiver poder suficiente.

````python
# The player's power starts out at 5.
power = 5

# The player is allowed to keep playing as long as their power is over 0.
while power > 0:
    print("You are still playing, because your power is %d." % power)
    # Your game code would go here, which includes challenges that make it
    #   possible to lose power.
    # We can represent that by just taking away from the power.
    power = power - 1

print("\nOh no, your power dropped to 0! Game Over.")
````

## Loops "Infinitos" e Break

Um loop infinito é executado indefinidamente porque sua condição sempre permanece True.
Geralmente são considerados erros, mas podem ser usados intencionalmente em alguns casos.
A instrução **break** encerra o loop atual e pula para a instrução imediatamente após o loop.

````python
n = 1
while n > 0:  # This creates an infinite loop
    print('Python')
    print('C++')
    n += 1

    if n > 3:
        break  # This will break out of the loop when count is greater than 2

print('Hey stop the loop')
````

## **Usando continue**

A instrução continue pula o restante da iteração atual do loop e passa para a próxima iteração.
Útil para pular certas iterações com base em uma condição.

**Exemplo:**

````python
#Print numbers from 1 to 10, except 4 and 7.
num = 1

while num <= 10:

  if num == 4 or num == 7:
    num += 1
    continue

  print(num)
  num+=1





````

## **Loop While com if-else**

Um loop while pode ter um bloco if-else opcional.
* Usamos a instrução if-else no loop quando é necessária iteração condicional. Ou seja, se a condição for True, as instruções dentro do bloco if serão executadas; caso contrário, o bloco else será executado.

* Se o loop for concluído sem encontrar uma instrução break, o código no bloco else será executado.
* Se uma instrução break for encontrada dentro do loop while, o bloco else será ignorado.

**Enumerate**
Você pode obter o mesmo efeito acima usando a função enumerate(), para obter dois valores a cada passo do loop.

````python
# the enumerate function will return two values each time the loop statement is run:
# "indx" will be the current index value, and "val" will be the current element of my_list
for indx, val in enumerate(fruits):
    print(indx, val)
````

## Padrões de Loop (contagem, soma, média, etc.)

Loops podem ser usados para realizar várias operações em dados.
Padrões comuns incluem contar itens, somar valores, calcular médias, encontrar valores máximos/mínimos, etc.

````python
#Counting - Count how many times the number '2' appears in a list:
numbers = [1, 2, 3, 2, 4, 2, 5]
count = 0
for num in numbers:
    if num == 2:
        count += 1
print(count)
````

````python
#Summing - Calculate the sum of all numbers in a list:

numbers = [1, 2, 3, 4, 5]
total = 0
for num in numbers:
    total += num
    print(total)
print(total)
````

Média - Calcular a média dos números em uma lista:

````python
numbers = [1, 2, 3, 4, 5]
total = 0
count = 0
for num in numbers:
   total += num
   count += 1

print('Exit loop')
average = total / count
print(average)

#alternative code
#total = sum(numbers)
#average = total / len(numbers)
````

````python
total = sum(numbers)
average = total / len(numbers)

print(average)
````
